# Macao Residential Property Price Index: Q3 Comparison, 2020–2025

## Research question
How did Macao's overall residential property price index change between the third quarters of 2020 and 2025?

## Data source and scope
The values below are transcribed from the Macao Statistics and Census Service (DSEC), *Residential Property Price Index, July–September 2025*, which includes a table comparing third-quarter values for past years. The index uses 2011 = 100. This is a six-point descriptive time-series exercise, not a measure of household affordability.

Official source: https://www.dsec.gov.mo/getAttachment/bde05008-86a7-4395-a616-c5c54ad617bb/E_IPH_FR_2025_M09.aspx

## Reproducibility
Run the cells from top to bottom. Required packages are listed in `requirements.txt`.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'year': [2020, 2021, 2022, 2023, 2024, 2025],
    'q3_property_price_index': [269.0, 269.3, 250.4, 243.0, 211.0, 191.6],
}
df = pd.DataFrame(data)
df['year_on_year_change_percent'] = df['q3_property_price_index'].pct_change() * 100
df
```

## Visualization 1: Index trend
The index was nearly unchanged from Q3 2020 to Q3 2021, then declined in each subsequent year shown. By Q3 2025, it was lower than the Q3 2020 value. The index measures price movement relative to its base year; it is not a house price in currency.

```python
plt.figure(figsize=(8, 4.8))
plt.plot(df['year'], df['q3_property_price_index'], marker='o')
plt.title('Macao Residential Property Price Index (Q3), 2020–2025')
plt.xlabel('Year (third quarter)')
plt.ylabel('Index (2011 = 100)')
plt.xticks(df['year'])
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Visualization 2: Change from the previous year's Q3
This chart calculates the percentage change between consecutive third-quarter observations. Negative values indicate that the index was lower than in the same quarter of the previous year. The first year has no change value because there is no earlier observation in this dataset.

```python
change_df = df.dropna(subset=['year_on_year_change_percent'])
plt.figure(figsize=(8, 4.8))
plt.bar(change_df['year'], change_df['year_on_year_change_percent'])
plt.axhline(0, linewidth=1)
plt.title('Year-on-Year Change in Q3 Property Price Index')
plt.xlabel('Year')
plt.ylabel("Change from previous year's Q3 (%)")
plt.xticks(change_df['year'])
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Key findings
- The index rose only slightly from 269.0 in Q3 2020 to 269.3 in Q3 2021.
- It then declined across the subsequent Q3 observations, reaching 191.6 in Q3 2025.
- The Q3 2025 index was about 28.8% lower than the Q3 2020 index (calculated from the two index values).

## One limitation
The dataset contains only six third-quarter observations of an overall price index. It does not include household income, mortgage costs, transaction-level prices, or differences between property types. Therefore, it cannot by itself determine whether housing became more affordable or explain why the index changed.
